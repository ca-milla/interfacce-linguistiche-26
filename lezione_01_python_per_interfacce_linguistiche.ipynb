## Stringhe: il primo dato linguistico
Una stringa è una sequenza ordinata di caratteri. Per un corso NLP è un primo modello importante: il testo non è ancora "significato", ma è già un oggetto che possiamo misurare e trasformare.

Le stringhe sono **immutabili**: `.upper()`, `.lower()`, `.strip()` e `.replace()` restituiscono una nuova stringa; non cambiano quella originale. Gli indici partono da `0`, quindi il primo carattere è `testo[0]`; `-1` indica l'ultimo carattere.

Lo slicing segue la forma `testo[inizio:fine]`: l'estremo finale è escluso. Questa convenzione è utile perché `testo[:3]` contiene esattamente i primi tre caratteri e perché gli intervalli si possono concatenare senza sovrapporsi.

Le virgolette singole e doppie hanno lo stesso ruolo. Le f-string permettono di inserire valori dentro una frase: `f"Token: {parola}"`. Attenzione però a distinguere il carattere spazio dalla stringa vuota `""`: il primo occupa una posizione, la seconda no.

```python
frase = "Il gatto dorme sul divano."
print(frase[0])
print(frase[-1])
print(frase[3:8])
print(frase.upper())
print(frase.lower())
print(len(frase))
print("gatto" in frase.lower())
```

```python
testo = "  Una frase con spazi.  "
print(testo.strip())
print(testo.replace("frase", "stringa"))
print("uno,due,tre".split(","))
print("uno  due".split())
print("uno  due".split(" "))
print("testo testo".count("testo"))
print("linguistica".index("g"))
parole = ["uno", "due", "tre"]
print(" - ".join(parole))
```

### Due strumenti per osservare il testo
La seconda lezione introduce due operazioni molto utili anche per dati linguistici:

- `.count(sottostringa)` conta quante volte compare una sequenza;
- `.index(sottostringa)` restituisce la posizione della prima occorrenza.

`split()` senza argomenti tratta una o più spaziature consecutive come separatori e ignora gli spazi ai bordi. `split(" ")`, invece, divide esattamente sul carattere spazio: con più spazi può produrre elementi vuoti. Per tokenizzare una frase semplice, `split()` è di solito la scelta più robusta.

Come sempre, `.count()` restituisce un numero, mentre `.index()` cerca una posizione. Se la sottostringa non esiste, `.index()` produce un errore: è un buon esempio del motivo per cui conviene provare anche casi limite.

## Condizioni e cicli
L'indentazione definisce il blocco. In Python, dopo `:` le istruzioni indentate appartengono a quella condizione o a quel ciclo. La convenzione è usare quattro spazi per livello e mantenere la stessa indentazione in tutto il blocco.

`for` attraversa gli elementi di una sequenza; `if` permette di scegliere un comportamento. `range(n)` produce i numeri da `0` a `n-1`.

Per chi arriva da JavaScript: le parentesi graffe non sono necessarie, ma gli spazi iniziali non sono semplicemente estetici. Un'indentazione incoerente può produrre `IndentationError` oppure, peggio, cambiare il blocco a cui appartiene un'istruzione.

```python
parole = ["il", "gatto", "dorme", "il", "gatto"]
for parola in parole:
    if len(parola) > 3:
        print(f"{parola}: parola lunga")
    else:
        print(f"{parola}: parola breve")

for indice in range(3):
    print(indice)
```

### Esercizio 10 - trasformare un filtro in funzione
Scrivete `filtra_parole(parole, lunghezza_minima)` che restituisca una nuova lista contenente solo le parole con almeno `lunghezza_minima` caratteri.

Provatela con la lista `parole` e con `lunghezza_minima` uguale a 3 e poi a 5. Usate un ciclo `for`, una condizione `if` e `.append()`: l'obiettivo è trasformare il procedimento appena visto in una funzione riutilizzabile.

```python
def filtra_parole(parole, lunghezza_minima):
    risultato = []
    for parola in parole:
        if len(parola) >= lunghezza_minima:
            risultato.append(parola)
    return risultato

print(filtra_parole(parole, 3))
print(filtra_parole(parole, 5))
```

## Dalla frase ai conteggi
Nei capitoli introduttivi di Jurafsky e Martin, una parte del lavoro NLP consiste nel trasformare una sequenza di testo in unità osservabili. Qui costruiamo una versione didattica e semplice: non è un tokenizer linguistico completo, ma rende visibili le decisioni.

La nostra pipeline è:
1. **normalizzare**: uniformare maiuscole e rimuovere alcuni segni;
2. **tokenizzare**: dividere la stringa in una lista con `.split()`;
3. **contare**: attraversare la lista e registrare quante volte compare ogni parola.

Ogni scelta ha conseguenze. Se rimuoviamo il punto, `"dorme."` diventa `"dorme"`; se non rimuoviamo l'apostrofo, forme come `l'analisi` restano un singolo token. In un progetto reale dovremmo decidere le regole in base alla domanda linguistica, non cercare una pulizia universale.

```python
testo = "Il gatto dorme. Il gatto sogna pesci e il cane corre!"

def normalizza(testo):
    return testo.lower().replace(".", "").replace("!", "")

testo_pulito = normalizza(testo)
token = testo_pulito.split()
print(testo_pulito)
print(token)
print(f"Numero di token: {len(token)}")
```

## Bigrammi: contesto locale
Una parola isolata dice poco sul contesto. Un **bigramma** è una coppia di token consecutivi: nella lista `["il", "gatto", "dorme"]` otteniamo `("il", "gatto")` e `("gatto", "dorme")`.

Questo prepara il terreno ai modelli n-gram di Jurafsky e Martin: un modello può descrivere quali parole tendono a comparire vicine. Qui non stimiamo ancora probabilità; impariamo solo a costruire le coppie e a controllare quante sono.

Regola utile: con `n` token ci sono `n - 1` bigrammi, se `n` è almeno 1. Il risultato è vuoto per una lista con zero o un solo token.

```python
def bigrammi(token):
    return [(token[indice], token[indice + 1]) for indice in range(len(token) - 1)]

coppie = bigrammi(token)
print(coppie)
print(f"Numero di bigrammi: {len(coppie)}")
```

## Esercizio finale - una funzione di analisi
Costruite `riassunto_testo(testo)` seguendo questa scaletta:
1. normalizzate il testo e dividetelo in token;
2. stampate il testo pulito e la lista dei token;
3. restituite una lista con tre elementi: la lista dei token, il numero di token e la lista delle frequenze;
4. provate la funzione su una frase scelta da voi;
5. controllate se il risultato risponde davvero alle domande che volevate fare.

Lavorate per passi: prima fate funzionare normalizzazione e `.split()`, poi aggiungete il conteggio con liste e `.count()`, infine organizzate il risultato in una lista. Se volete una sfida ulteriore, aggiungete anche i bigrammi, usando solo le funzioni già scritte.

Domande di discussione: quali casi reali non gestisce il nostro tokenizer? Pensate ad apostrofi, punteggiatura, emoji, parole composte e maiuscole. In che modo una scelta di tokenizzazione può cambiare un conteggio?

```python
def riassunto_testo(testo):
    testo_pulito = normalizza(testo)
    token = testo_pulito.split()
    frequenze = []
    parole_contate = []
    for parola in token:
        if parola not in parole_contate:
            frequenze.append(f"{parola}: {token.count(parola)}")
            parole_contate.append(parola)
    return [token, len(token), frequenze]

riassunto = riassunto_testo("Il cane corre e il cane gioca.")
print(riassunto)
```
